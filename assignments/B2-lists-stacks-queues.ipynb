```python
class Node:
    def __init__(self, data, next=None):
        self.data = data
        self.next = next


def from_list(items):
    head = None
    for value in reversed(items):
        head = Node(value, head)
    return head


def to_list(head):
    out = []
    while head is not None:
        out.append(head.data)
        head = head.next
    return out
```

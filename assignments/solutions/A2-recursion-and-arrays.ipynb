## Setup

Run this cell once. It provides `check_recursive`, which the checks use.

```python
import inspect
import re


def check_recursive(fn, allow_loops=False):
    """Passes when `fn` calls itself and, unless allow_loops=True, uses no for/while statements."""
    lines = inspect.getsource(fn).split("\n")[1:]
    body = "\n".join(lines)
    assert re.search(rf"\b{fn.__name__}\s*\(", body), f"{fn.__name__} should call itself (recursion)"
    if not allow_loops:
        assert not re.search(r"^\s*(for|while)\b", body, re.M), "No loops here: use recursion instead"
```

## Problem 1: Maximum of a list

Write `max_of_list(nums)` for a non-empty list, without using `max()`.

```python
def max_of_list(nums):
    if len(nums) == 1:
        return nums[0]
    rest = max_of_list(nums[1:])
    return nums[0] if nums[0] > rest else rest
```

```python
assert max_of_list([3, 9, 2]) == 9
assert max_of_list([5]) == 5
assert max_of_list([-4, -2, -9]) == -2
assert max_of_list([1, 2, 3, 4, 5]) == 5
check_recursive(max_of_list)
print('✅ Problem 1 passed')
```

## Problem 2: Count occurrences

Write `count_occurrences(nums, x)`: how many times does `x` appear in `nums`? An empty list gives 0.

```python
def count_occurrences(nums, x):
    if not nums:
        return 0
    return (1 if nums[0] == x else 0) + count_occurrences(nums[1:], x)
```

```python
assert count_occurrences([], 3) == 0
assert count_occurrences([2, 2, 5, 2], 2) == 3
assert count_occurrences([1, 2, 3], 9) == 0
check_recursive(count_occurrences)
print('✅ Problem 2 passed')
```

## Problem 3: Last index

Write `last_index(nums, x, i=None)` returning the last index of `x`, or -1 if it is absent. Start from the end.

```python
def last_index(nums, x, i=None):
    if i is None:
        i = len(nums) - 1
    if i < 0:
        return -1
    if nums[i] == x:
        return i
    return last_index(nums, x, i - 1)
```

```python
assert last_index([3, 2, 5, 2, 8, 2, 1], 2) == 5
assert last_index([3, 2, 5], 9) == -1
assert last_index([], 1) == -1
assert last_index([7], 7) == 0
check_recursive(last_index)
print('✅ Problem 3 passed')
```

## Problem 4: Return a new list of indices

Write `indices_greater_than(nums, k, i=0)` returning a **new list** of the indices whose value is greater than `k` (in increasing order).

```python
def indices_greater_than(nums, k, i=0):
    if i == len(nums):
        return []
    rest = indices_greater_than(nums, k, i + 1)
    return ([i] if nums[i] > k else []) + rest
```

```python
assert indices_greater_than([3, 8, 1, 9, 5], 4) == [1, 3, 4]
assert indices_greater_than([], 0) == []
assert indices_greater_than([1, 1], 5) == []
check_recursive(indices_greater_than)
print('✅ Problem 4 passed')
```

## Problem 5: Reverse a list

Write `reverse_list(nums)` returning a new reversed list. The original list must not change.

```python
def reverse_list(nums):
    if not nums:
        return []
    return reverse_list(nums[1:]) + [nums[0]]
```

```python
data = [1, 2, 3]
assert reverse_list(data) == [3, 2, 1]
assert data == [1, 2, 3]
assert reverse_list([]) == []
assert reverse_list([7]) == [7]
check_recursive(reverse_list)
print('✅ Problem 5 passed')
```

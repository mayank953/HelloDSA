## Setup

Run this cell once. It prepares what the problems and their checks need (for example the `check_recursive` helper or a `Node` class).

```python
import inspect
import re


def check_recursive(fn, allow_loops=False):
    """Passes when `fn` calls itself and, unless allow_loops=True, uses no for/while statements."""
    lines = inspect.getsource(fn).split("\n")[1:]
    body = "\n".join(lines)
    assert re.search(rf"\b{fn.__name__}\s*\(", body), f"{fn.__name__} should call itself (recursion)"
    if not allow_loops:
        assert not re.search(r"^\s*(for|while)\b", body, re.M), "No loops here: use recursion instead"
```

## Problem 1: Find the first occurrence

Write `find_first(nums, x, i=0)` returning the index of the first `x`, or -1.

```python
def find_first(nums, x, i=0):
    if i == len(nums):
        return -1
    if nums[i] == x:
        return i
    return find_first(nums, x, i + 1)
```

```python
assert find_first([4, 7, 7, 1], 7) == 1
assert find_first([4, 7], 9) == -1
assert find_first([], 1) == -1
check_recursive(find_first)
print('✅ Problem 1 passed')
```

## Problem 2: Binary search

Write `binary_search(nums, target, low=0, high=None)` for a sorted list. Return the index or -1.

```python
def binary_search(nums, target, low=0, high=None):
    if high is None:
        high = len(nums) - 1
    if low > high:
        return -1
    mid = (low + high) // 2
    if nums[mid] == target:
        return mid
    if nums[mid] < target:
        return binary_search(nums, target, mid + 1, high)
    return binary_search(nums, target, low, mid - 1)
```

```python
data = [1, 3, 5, 7, 9, 11]
assert [binary_search(data, v) for v in data] == [0, 1, 2, 3, 4, 5]
assert binary_search(data, 4) == -1
assert binary_search(data, 100) == -1
assert binary_search([], 1) == -1
check_recursive(binary_search)
print('✅ Problem 2 passed')
```

## Problem 3: Merge two sorted lists

Write `merge(a, b)` that merges two sorted lists into one new sorted list. You can use a loop or recursion.

```python
def merge(a, b):
    out = []
    i = j = 0
    while i < len(a) and j < len(b):
        if a[i] <= b[j]:
            out.append(a[i])
            i += 1
        else:
            out.append(b[j])
            j += 1
    return out + a[i:] + b[j:]
```

```python
assert merge([1, 4, 9], [2, 3, 10]) == [1, 2, 3, 4, 9, 10]
assert merge([], [1]) == [1]
assert merge([1, 1], [1]) == [1, 1, 1]
assert merge([], []) == []
print('✅ Problem 3 passed')
```

## Problem 4: Merge sort

Write `merge_sort(nums)` returning a new sorted list. Use your `merge` from the previous problem.

```python
def merge_sort(nums):
    if len(nums) <= 1:
        return nums[:]
    mid = len(nums) // 2
    return merge(merge_sort(nums[:mid]), merge_sort(nums[mid:]))
```

```python
import random
random.seed(1)
for _ in range(30):
    data = [random.randint(-20, 20) for _ in range(random.randint(0, 12))]
    copy = data[:]
    assert merge_sort(data) == sorted(data)
    assert data == copy
check_recursive(merge_sort)
print('✅ Problem 4 passed')
```

## Problem 5: Quick sort

Write `quick_sort(nums)` returning a new sorted list. Use the last element as the pivot. It must work with duplicates.

```python
def quick_sort(nums):
    if len(nums) <= 1:
        return nums[:]
    pivot = nums[-1]
    smaller = [x for x in nums[:-1] if x <= pivot]
    larger = [x for x in nums[:-1] if x > pivot]
    return quick_sort(smaller) + [pivot] + quick_sort(larger)
```

```python
import random
random.seed(2)
for _ in range(30):
    data = [random.randint(-5, 5) for _ in range(random.randint(0, 15))]
    copy = data[:]
    assert quick_sort(data) == sorted(data)
    assert data == copy
check_recursive(quick_sort)
print('✅ Problem 5 passed')
```

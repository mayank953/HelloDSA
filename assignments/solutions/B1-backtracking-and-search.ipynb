## Problem 1: Generate parentheses

Write `generate_parentheses(n)` returning every valid string of `n` pairs of brackets, in any order.

```python
def generate_parentheses(n):
    answers = []

    def explore(current, opened, closed):
        if len(current) == 2 * n:
            answers.append(current)
            return
        if opened < n:
            explore(current + '(', opened + 1, closed)
        if closed < opened:
            explore(current + ')', opened, closed + 1)

    explore('', 0, 0)
    return answers
```

```python
assert sorted(generate_parentheses(3)) == sorted(['((()))', '(()())', '(())()', '()(())', '()()()'])
assert [len(generate_parentheses(n)) for n in range(1, 7)] == [1, 2, 5, 14, 42, 132]
assert generate_parentheses(0) == ['']
print('✅ Problem 1 passed')
```

## Problem 2: Count the N-Queens solutions

Write `count_n_queens(n)`: in how many ways can `n` queens be placed on an `n x n` board so that none attack each other?

```python
def count_n_queens(n):
    used_cols, diag_down, diag_up = set(), set(), set()

    def place(row):
        if row == n:
            return 1
        total = 0
        for col in range(n):
            if col in used_cols or (row - col) in diag_down or (row + col) in diag_up:
                continue
            used_cols.add(col); diag_down.add(row - col); diag_up.add(row + col)
            total += place(row + 1)
            used_cols.remove(col); diag_down.remove(row - col); diag_up.remove(row + col)
        return total

    return place(0)
```

```python
assert [count_n_queens(n) for n in range(1, 8)] == [1, 0, 0, 2, 10, 4, 40]
assert count_n_queens(8) == 92
print('✅ Problem 2 passed')
```

## Problem 3: Word search

Write `word_exists(board, word)`: can `word` be traced through adjacent cells (up, down, left, right) of `board` (a list of lists of letters), using each cell at most once? The board must be left unchanged.

```python
def word_exists(board, word):
    rows, cols = len(board), len(board[0])

    def explore(r, c, k):
        if k == len(word):
            return True
        if not (0 <= r < rows and 0 <= c < cols) or board[r][c] != word[k]:
            return False
        saved, board[r][c] = board[r][c], '#'
        found = any(explore(r + dr, c + dc, k + 1) for dr, dc in [(1, 0), (-1, 0), (0, 1), (0, -1)])
        board[r][c] = saved
        return found

    return any(explore(r, c, 0) for r in range(rows) for c in range(cols))
```

```python
grid = [list('ABCE'), list('SFCS'), list('ADEE')]
assert word_exists(grid, 'ABCCED') and word_exists(grid, 'SEE') and not word_exists(grid, 'ABCB')
assert grid == [list('ABCE'), list('SFCS'), list('ADEE')]
print('✅ Problem 3 passed')
```

## Problem 4: The binary search template

Write `first_true(low, high, is_true)`: the smallest `x` in `[low, high]` for which the monotone question `is_true(x)` is `True`, or `high + 1` if there is none.

```python
def first_true(low, high, is_true):
    answer = high + 1
    while low <= high:
        mid = low + (high - low) // 2
        if is_true(mid):
            answer = mid
            high = mid - 1
        else:
            low = mid + 1
    return answer
```

```python
assert first_true(0, 100, lambda x: x * x >= 50) == 8
assert first_true(0, 5, lambda x: False) == 6
assert first_true(0, 5, lambda x: True) == 0
assert first_true(5, 5, lambda x: True) == 5 and first_true(5, 5, lambda x: False) == 6
print('✅ Problem 4 passed')
```

## Problem 5: Slowest eating speed

Piles of bananas must be eaten within `hours`. Each hour you pick one pile and eat up to `speed` bananas from it. Write `min_eating_speed(piles, hours)` returning the smallest integer speed that works. Hint: binary search on the answer.

```python
def min_eating_speed(piles, hours):
    low, high, answer = 1, max(piles), max(piles)
    while low <= high:
        mid = (low + high) // 2
        needed = sum(-(-pile // mid) for pile in piles)
        if needed <= hours:
            answer = mid
            high = mid - 1
        else:
            low = mid + 1
    return answer
```

```python
assert min_eating_speed([3, 6, 7, 11], 8) == 4
assert min_eating_speed([30, 11, 23, 4, 20], 5) == 30
assert min_eating_speed([30, 11, 23, 4, 20], 6) == 23
assert min_eating_speed([1], 1) == 1
print('✅ Problem 5 passed')
```

## Problem 6: Search in a rotated array

A sorted list of distinct numbers was rotated, for example `[4, 5, 6, 7, 0, 1, 2]`. Write `search_rotated(items, target)` returning the index of `target` or -1, in O(log n).

```python
def search_rotated(items, target):
    low, high = 0, len(items) - 1
    while low <= high:
        mid = low + (high - low) // 2
        if items[mid] == target:
            return mid
        if items[low] <= items[mid]:
            if items[low] <= target < items[mid]:
                high = mid - 1
            else:
                low = mid + 1
        else:
            if items[mid] < target <= items[high]:
                low = mid + 1
            else:
                high = mid - 1
    return -1
```

```python
base = [10, 20, 30, 40, 50, 60, 70]
for k in range(len(base)):
    rotated = base[k:] + base[:k]
    assert [search_rotated(rotated, v) for v in base] == [rotated.index(v) for v in base]
    assert search_rotated(rotated, 35) == -1
assert search_rotated([], 1) == -1
print('✅ Problem 6 passed')
```

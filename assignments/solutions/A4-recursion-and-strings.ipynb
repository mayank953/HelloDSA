## Setup

Run this cell once. It provides `check_recursive`, which the checks use.

```python
import inspect
import re


def check_recursive(fn, allow_loops=False):
    """Passes when `fn` calls itself and, unless allow_loops=True, uses no for/while statements."""
    lines = inspect.getsource(fn).split("\n")[1:]
    body = "\n".join(lines)
    assert re.search(rf"\b{fn.__name__}\s*\(", body), f"{fn.__name__} should call itself (recursion)"
    if not allow_loops:
        assert not re.search(r"^\s*(for|while)\b", body, re.M), "No loops here: use recursion instead"
```

## Problem 1: Remove a character

Write `remove_char(s, ch)` returning `s` without any `ch`.

```python
def remove_char(s, ch):
    if s == '':
        return ''
    rest = remove_char(s[1:], ch)
    return rest if s[0] == ch else s[0] + rest
```

```python
assert remove_char('good morning', 'o') == 'gd mrning'
assert remove_char('zzz', 'z') == ''
assert remove_char('', 'a') == ''
assert remove_char('abc', 'x') == 'abc'
check_recursive(remove_char)
print('✅ Problem 1 passed')
```

## Problem 2: Reverse a string

Write `reverse_string(s)` recursively.

```python
def reverse_string(s):
    if len(s) <= 1:
        return s
    return reverse_string(s[1:]) + s[0]
```

```python
assert reverse_string('abc') == 'cba'
assert reverse_string('') == ''
assert reverse_string('a') == 'a'
assert reverse_string('racecar') == 'racecar'
check_recursive(reverse_string)
print('✅ Problem 2 passed')
```

## Problem 3: All subsequences

Write `subsequences(s)` returning a list of all subsequences of `s`, including the empty string, in any order.

```python
def subsequences(s):
    if s == '':
        return ['']
    rest = subsequences(s[1:])
    return rest + [s[0] + x for x in rest]
```

```python
assert sorted(subsequences('abc')) == sorted(['', 'a', 'b', 'c', 'ab', 'ac', 'bc', 'abc'])
assert subsequences('') == ['']
assert len(subsequences('abcd')) == 16
check_recursive(subsequences)
print('✅ Problem 3 passed')
```

## Problem 4: All permutations

Write `permutations(s)` returning a list of every permutation of a string with distinct letters, in any order.

```python
def permutations(s):
    if len(s) <= 1:
        return [s]
    out = []
    for i, ch in enumerate(s):
        for p in permutations(s[:i] + s[i + 1:]):
            out.append(ch + p)
    return out
```

```python
from itertools import permutations as _perms
assert sorted(permutations('abc')) == sorted(''.join(p) for p in _perms('abc'))
assert len(permutations('abcd')) == 24
assert permutations('a') == ['a']
check_recursive(permutations, allow_loops=True)
print('✅ Problem 4 passed')
```

## Problem 5: Phone keypad words

The keypad mapping is given. Write `letter_combinations(digits)` returning every word the digit string could spell (`['']` for an empty string).

```python
KEYS = {'2': 'abc', '3': 'def', '4': 'ghi', '5': 'jkl', '6': 'mno', '7': 'pqrs', '8': 'tuv', '9': 'wxyz'}


def letter_combinations(digits):
    if digits == '':
        return ['']
    rest = letter_combinations(digits[1:])
    return [c + word for c in KEYS[digits[0]] for word in rest]
```

```python
assert letter_combinations('') == ['']
assert sorted(letter_combinations('7')) == ['p', 'q', 'r', 's']
assert sorted(letter_combinations('23')) == sorted(a + b for a in 'abc' for b in 'def')
assert len(letter_combinations('92')) == 12
check_recursive(letter_combinations, allow_loops=True)
print('✅ Problem 5 passed')
```

## Problem 6: Tower of Hanoi moves

Write `hanoi_moves(n, source='A', target='C', spare='B')` returning the list of moves as `(from_peg, to_peg)` tuples that moves `n` disks from `source` to `target`.

```python
def hanoi_moves(n, source='A', target='C', spare='B'):
    if n == 0:
        return []
    return (hanoi_moves(n - 1, source, spare, target)
            + [(source, target)]
            + hanoi_moves(n - 1, spare, target, source))
```

```python
for n in range(0, 6):
    moves = hanoi_moves(n)
    assert len(moves) == 2 ** n - 1
    pegs = {'A': list(range(n, 0, -1)), 'B': [], 'C': []}
    for a, b in moves:
        disk = pegs[a].pop()
        assert not pegs[b] or pegs[b][-1] > disk, 'a larger disk was placed on a smaller one'
        pegs[b].append(disk)
    assert pegs['C'] == list(range(n, 0, -1)), 'all disks must end on the target peg'
check_recursive(hanoi_moves)
print('✅ Problem 6 passed')
```

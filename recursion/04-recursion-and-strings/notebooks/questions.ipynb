# Remove Z

    """
    Remove all occurrences of 'Z' and 'z' from the given string.
    
    Example:
    Input: "ZebRa"
    Output: "ebRa"
    
    Input: "ZZZzzzZZZ"
    Output: ""
    """

```python
def removeZ(string):
    if(string==""):
        return string

    if(string[0] in 'zZ'):
        return removeZ(string[1:])
    return string[0] + removeZ(string[1:])

removeZ("ZzA")
```

# Palindrome

```python

def is_palindrome_helper(s,start,end):
    if(start>=end):
        return True

    return s[start]==s[end] and is_palindrome_helper(s,start+1,end-1)

def is_palindrome(s: str) -> bool:
    if(len(s) == 0 or len(s)==1):
        return True

    
    # s = s.lower()
    s = "".join(c for c in s if c.isalnum()).lower()


    return is_palindrome_helper(s,0,len(s)-1)

def test_is_palindrome():
    assert is_palindrome("A man a plan a canal Panama") == True, "Test case 1 failed"
    assert is_palindrome("race a car") == False, "Test case 2 failed"
    assert is_palindrome("Was it a car or a cat I saw?") == True, "Test case 3 failed"
    print("All test cases for is_palindrome passed!")



test_is_palindrome()
```

```python
s = 'abc'
```

```python
def print_subsequences(input,output):
    if(input==''):
        print(output)
        return
        
    print_subsequences(input[1:],output)

    print_subsequences(input[1:],output+input[0])


print_subsequences(s,"")
```

```python
def return_subsequences(input):
    ans = []
    if(input==""):
        ans.append("")
        return ans

    shortAnswer = return_subsequences(input[1:])

    ans.extend(shortAnswer)

    for i in shortAnswer:
        ans.append(input[0]+i)

    return ans
```

# Permutations

```python
def permutations(input,output):
    if(input==""):
        print(output)
        return

    first = input[0]
    for i in range(0,len(output)+1):
        permutations(input[1:],output[0:i]+first + output[i:])
    
```

```python
def permutations_returnList(input,output):
    
    if(input==""):
        ans = []
        ans.append(output)
        return ans
    ans = []
    first = input[0]
    for i in range(0,len(output)+1):
        ans.extend(permutations_returnList(input[1:],output[0:i]+first + output[i:]))

    return ans
    
```

```python
def permutations_returnList(input):
    ans = []
    if(input==""):
        ans = [""]
        return ans
        
    for i in range(0,len(input)):
        small_permutations = permutations_returnList(input[0:i] + input[i+1:])
        for each_permutation in small_permutations:
            ans.append(input[i] + each_permutation)

    return ans


```

```python
def tower_of_hanoi(n,source,extra,destination):
    if(n==0):
        return
    if(n==1):
        print(source,destination)
        return
    if(n==2):
        print(source,extra)
        print(source,destination)
        print(extra,destination)
        return

    tower_of_hanoi(n-1,source,destination,extra)
    print(source,destination)
    tower_of_hanoi(n-1,extra,source,destination)
```

```python
tower_of_hanoi(3,'a','b','c')
```

```python
keys = {'1':'','2':'abc','3':'def','4':'ghi','5':'jkl','6':'mno','7':'pqrs','8':'tuv','9':'wxyz'}
```

```python
def print_all_combination(input,output):
    if(len(input)==0):
        print(output)
        return

    s1 = keys[input[0]]
    for i in range(len(s1)):
        print_all_combination(input[1:],output+s1[i])
    
```

```python
print_all_combination('23','')
```

```python
def return_all_combination(input,output):
    
    ans =[]
    if(len(input)==0):
        ans.append(output)
        return ans

    s1 = keys[input[0]]
    for i in range(len(s1)):
        ans.extend(return_all_combination(input[1:],output+s1[i]))

    return ans
    
```

```python
def return_all_combination_without_output(input):
    
    ans =[]
    if(len(input)==0):
        return ['']

    s1 = keys[input[0]]

    sans = return_all_combination_without_output(input[1:])
    # print(s1,sans)
    for each in s1:
        for chottaAnswer in sans:
            ans.append(each+chottaAnswer)

    return ans
    
```

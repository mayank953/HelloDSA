# Stack Using List

```python
class StackUsingList:
    def __init__(self):
        self.stack = []  # Initialize an empty list to represent the stack

    def is_empty(self):
        return len(self.stack) == 0  # Stack is empty if its length is 0

    def push(self, data):
        self.stack.append(data)  # Append the new data to the end of the list
        print(f"Pushed {data} onto the stack.")

    def pop(self):
        if self.is_empty():
            print("Stack underflow. Cannot pop from an empty stack.")
            return None
        return self.stack.pop()  # Remove and return the last element (top of the stack)

    def peek(self):
        if self.is_empty():
            print("Stack is empty. No top element.")
            return None
        return self.stack[-1]  # Return the last element (top of the stack) without removing it
```

# Stack using array

```python
class StackUsingArray:
    def __init__(self, capacity):
        self.capacity = capacity  # Set the maximum capacity of the stack
        self.stack = [None] * capacity  # Initialize an array of fixed size
        self.top = -1  # Initialize top to -1, meaning the stack is empty

    def is_empty(self):
        return self.top == -1  # Stack is empty if top is -1

    def is_full(self):
        return self.top == self.capacity - 1  # Stack is full if top reaches capacity - 1

    def push(self, data):
        if self.is_full():
            print("Stack overflow. Cannot push the value.")
            return
        self.top += 1  # Increment top index
        self.stack[self.top] = data  # Place the new item at the top
        print(f"Pushed {data} onto the stack.")

    def pop(self):
        if self.is_empty():
            print("Stack underflow. Cannot pop from an empty stack.")
            return None
        popped_data = self.stack[self.top]  # Get the top item
        self.top -= 1  # Decrement top index
        return popped_data  # Return the popped value

    def peek(self):
        if self.is_empty():
            print("Stack is empty. No top element.")
            return None
        return self.stack[self.top]  # Return the current top element
```

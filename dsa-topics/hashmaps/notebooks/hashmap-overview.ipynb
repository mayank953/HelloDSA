```python
class HashMap:
    def __init__(self, size):
        """
        Initialize the HashMap with a given size.
        We use two arrays: 'slots' to store keys and 'values' to store corresponding values.
        :param size: Size of the hash table.
        """
        self.size = size
        self.slots = [None] * self.size  # To store keys
        self.values = [None] * self.size  # To store values

    def hash_function(self, key):
        """
        Hash function to calculate the index from the key.
        This uses Python's built-in hash function and then applies modulus to fit it into the table size.
        :param key: The key to hash.
        :return: Index within the table.
        """
        return abs(hash(key)) % self.size

    def rehash(self, old_hash):
        """
        Linear probing to find the next available slot.
        If a collision occurs, this function provides the next index by incrementing the old index.
        :param old_hash: Previous hash index.
        :return: New hash index after probing.
        """
        return (old_hash + 1) % self.size

    def put(self, key, value):
        """
        Insert key-value pair into the HashMap.
        If a collision occurs, it resolves it using linear probing.
        :param key: The key to insert.
        :param value: The value associated with the key.
        """
        hash_value = self.hash_function(key)

        # If slot is empty, insert the key-value pair
        if self.slots[hash_value] is None:
            self.slots[hash_value] = key
            self.values[hash_value] = value
        else:
            # If the key already exists, update its value
            if self.slots[hash_value] == key:
                self.values[hash_value] = value
            else:
                # Resolve collision with linear probing
                new_hash_value = self.rehash(hash_value)

                # Continue probing until an empty slot is found or the key is found , 2nd is very important
                while self.slots[new_hash_value] is not None and self.slots[new_hash_value] != key:
                    new_hash_value = self.rehash(new_hash_value)

                # If it's an empty slot, insert the key-value pair
                if self.slots[new_hash_value] is None:
                    self.slots[new_hash_value] = key
                    self.values[new_hash_value] = value
                else:
                    # If the key already exists in the table, update its value
                    self.values[new_hash_value] = value

    def get(self, key):
        """
        Retrieve the value associated with the given key.
        If the key does not exist, return None.
        :param key: The key to search for.
        :return: Value associated with the key or None if key is not found.
        """
        start_position = self.hash_function(key)
        current_position = start_position

        # Probe through the table to find the key
        while self.slots[current_position] is not None:
            if self.slots[current_position] == key:
                return self.values[current_position]  # Return the value if key is found
            
            # Move to the next slot (linear probing)
            current_position = self.rehash(current_position)

            # If we come back to the starting position, the key is not present
            if current_position == start_position:
                return None

        return None

    def delete(self, key):
        """
        Deletes the key-value pair from the HashMap.
        Uses linear probing to find the key and set it to None.
        :param key: The key to be deleted.
        """
        start_position = self.hash_function(key)
        current_position = start_position
    
        # Probe through the table to find the key
        while self.slots[current_position] is not None:
            if self.slots[current_position] == key:
                # Key found, set the slot and value to None
                self.slots[current_position] = None
                self.values[current_position] = None
                print(f"Key '{key}' deleted.")
                return
            
            # Move to the next slot (linear probing)
            current_position = self.rehash(current_position)
    
            # If we come back to the starting position, the key is not present
            if current_position == start_position:
                break
    
        print(f"Key '{key}' not found.")

    def __getitem__(self, key):
        """
        Allows usage of [] syntax to retrieve a value.
        E.g., hashmap[key] -> gets the value for the key.
        """
        return self.get(key)

    def __setitem__(self, key, value):
        """
        Allows usage of [] syntax to insert a key-value pair.
        E.g., hashmap[key] = value -> inserts the key-value pair.
        """
        self.put(key, value)

    def __str__(self):
        """
        String representation of the HashMap.
        Displays all non-empty slots with their key-value pairs.
        :return: String representation of the HashMap.
        """
        result = []
        for i in range(self.size):
            if self.slots[i] is not None:
                result.append(f"Slot {i}: {self.slots[i]} => {self.values[i]}")
            else:
                result.append(f"Slot {i}: Empty")
        return "\n".join(result)

# Example Usage:

# Create a HashMap with a fixed size
hash_map = HashMap(5)

# Insert key-value pairs
hash_map.put('apple', 10)
hash_map.put('banana', 20)
hash_map.put('orange', 30)

# Access values
print(hash_map.get('apple'))  # Output: 10
print(hash_map.get('banana'))  # Output: 20
print(hash_map.get('grape'))  # Output: None

# Insert and overwrite a key's value
hash_map.put('apple', 40)
print(hash_map.get('apple'))  # Output: 40

# Display the HashMap
print(hash_map)

# Output:
# Slot 0: Empty
# Slot 1: apple => 40
# Slot 2: Empty
# Slot 3: banana => 20
# Slot 4: orange => 30
```

```python


class Node:
    def __init__(self, key, value):
        """
        Initializes a node with a key-value pair.
        :param key: The key for the node.
        :param value: The value to store.
        """
        self.key = key
        self.value = value
        self.next = None  # Pointer to the next node
```

```python

# LinkedList and Node classes to be used in the HashMap for chaining.
class LinkedList:
    def __init__(self):
        """
        Initializes an empty linked list for handling collisions in HashMap buckets.
        """
        self.head = None

    def add(self, key, value):
        """
        Adds a new node with the given key-value pair at the head of the linked list.
        :param key: The key for the node.
        :param value: The value to store.
        """
        new_node = Node(key, value)
        new_node.next = self.head
        self.head = new_node

    def search(self, key):
        """
        Searches for a node with the given key.
        :param key: The key to search for.
        :return: The node if found, otherwise None.
        """
        current = self.head
        while current is not None:
            if current.key == key:
                return current
            current = current.next
        return None

    def remove(self, key):
        """
        Removes the node with the given key from the linked list.
        :param key: The key to remove.
        :return: True if the node was removed, False if not found.
        """
        current = self.head
        previous = None

        while current is not None:
            if current.key == key:
                if previous is None:  # Removing the head
                    self.head = current.next
                else:
                    previous.next = current.next
                return True
            previous = current
            current = current.next

        return False

    def traverse(self):
        """
        Prints the key-value pairs in the linked list.
        """
        current = self.head
        while current:
            print(f"{current.key}: {current.value} ->", end=" ")
            current = current.next
        print("None")
```

```python
class HashMap:
    def __init__(self, capacity=10):
        """
        Initializes the HashMap with the given capacity.
        Each bucket in the HashMap will store a linked list to handle collisions.
        :param capacity: The initial size of the HashMap.
        """
        self.capacity = capacity
        self.size = 0
        self.buckets = self._create_buckets(self.capacity)

    def _create_buckets(self, capacity):
        """
        Helper function to create an array of linked lists (buckets) for storing key-value pairs.
        :param capacity: The number of buckets.
        :return: A list of linked lists (LL).
        """
        return [LinkedList() for _ in range(capacity)]

    def hash_function(self, key):
        """
        Hashes the key using Python's built-in hash function and maps it to a bucket index.
        :param key: The key to hash.
        :return: The index of the bucket where the key should go.
        """
        return abs(hash(key)) % self.capacity

    def put(self, key, value):
        """
        Adds or updates a key-value pair in the HashMap.
        Uses chaining (LinkedList) to handle collisions.
        :param key: The key for the value.
        :param value: The value to store.
        """
        bucket_index = self.hash_function(key)
        bucket = self.buckets[bucket_index]

        node = bucket.search(key)
        if node is None:
            bucket.add(key, value)
            self.size += 1
        else:
            node.value = value  # Update existing key with new value

    def get(self, key):
        """
        Retrieves the value associated with the key.
        :param key: The key to search for.
        :return: The value if found, or None if not found.
        """
        bucket_index = self.hash_function(key)
        bucket = self.buckets[bucket_index]

        node = bucket.search(key)
        return node.value if node else "Key not found"

    def delete(self, key):
        """
        Removes the key-value pair from the HashMap.
        :param key: The key to remove.
        """
        bucket_index = self.hash_function(key)
        bucket = self.buckets[bucket_index]

        removed = bucket.remove(key)
        if removed:
            self.size -= 1
            print(f"Key '{key}' removed.")
        else:
            print(f"Key '{key}' not found.")

    def __len__(self):
        """
        Returns the number of elements in the HashMap.
        """
        return self.size

    def __str__(self):
        """
        Prints the content of the HashMap for visualization.
        """
        for i in range(self.capacity):
            print(f"Bucket {i}:", end=" ")
            self.buckets[i].traverse()
        return ""

```

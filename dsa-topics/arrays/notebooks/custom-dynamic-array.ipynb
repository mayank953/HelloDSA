```python
import ctypes

class CustomList:
    """
    A custom implementation of a dynamic array using C-types.
    Similar to Python's built-in list but created from scratch.
    """

    def __init__(self):
        """
        Initializes an empty list with a starting capacity of 1.
        """
        self.capacity = 1  # Initial capacity of the array
        self.size = 0  # Number of elements in the array
        self.array = self._create_array(self.capacity)  # Create the C-type array

    
    def _create_array(self, capacity):
        """
        Creates a new array with the given capacity using C-types.
        """
        return (capacity * ctypes.py_object)()  # C-type referential array

    def __len__(self):
        """
        Returns the number of elements currently in the list.
        """
        return self.size

    def append(self, item):
        """
        Adds a new item to the end of the array. Resizes the array if it's full.
        """
        if self.size == self.capacity:
            # If the array is full, double the capacity
            self._resize(2 * self.capacity)

        self.array[self.size] = item  # Add item at the current position
        self.size += 1  # Increment the size of the array

    def pop(self):
        """
        Removes and returns the last item in the array. Returns a message if empty.
        """
        if self.size == 0:
            return 'Empty List'
        
        popped_item = self.array[self.size - 1]  # Get the last item
        self.size -= 1  # Reduce size
        return popped_item

    def clear(self):
        """
        Clears the array, resets size and capacity.
        """
        self.size = 0
        self.capacity = 1
        self.array = self._create_array(self.capacity)  # Reset array

    def find(self, item):
        """
        Searches for the item in the array and returns its index if found.
        Returns a ValueError message if the item is not present.
        """
        for i in range(self.size):
            if self.array[i] == item:
                return i
        return 'ValueError - Item not found in list'

    def insert(self, pos, item):
        """
        Inserts an item at the given position. Resizes if the array is full.
        """
        if self.size == self.capacity:
            self._resize(2 * self.capacity)

        for i in range(self.size, pos, -1):
            self.array[i] = self.array[i - 1]  # Shift elements to the right

        self.array[pos] = item  # Insert the new item
        self.size += 1

    def remove(self, item):
        """
        Removes the first occurrence of the item from the list.
        If the item is not found, returns the error message from find().
        """
        pos = self.find(item)
        if isinstance(pos, int):  # If item is found
            self.__delitem__(pos)
        else:
            return pos  # If item is not found

    def __delitem__(self, pos):
        """
        Removes the item at the specified position and shifts the rest of the elements.
        """
        if 0 <= pos < self.size:
            for i in range(pos, self.size - 1):
                self.array[i] = self.array[i + 1]  # Shift elements to the left
            self.size -= 1
        else:
            raise IndexError("Index out of bounds")

    def __getitem__(self, index):
        """
        Returns the item at the specified index. Raises IndexError if out of bounds.
        """
        if 0 <= index < self.size:
            return self.array[index]
        else:
            raise IndexError('IndexError: List index out of range')

    def __str__(self):
        """
        Returns a string representation of the list.
        """
        elements = [str(self.array[i]) for i in range(self.size)]
        return "[" + ", ".join(elements) + "]"

    def _resize(self, new_capacity):
        """
        Resizes the array to a new capacity by creating a new array
        and copying the existing elements over.
        """
        new_array = self._create_array(new_capacity)  # Create a new array
        for i in range(self.size):
            new_array[i] = self.array[i]  # Copy elements to the new array

        self.array = new_array  # Replace the old array
        self.capacity = new_capacity  # Update the capacity

 
```

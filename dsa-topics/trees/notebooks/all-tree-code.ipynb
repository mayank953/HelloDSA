```python
class TreeNode:
    def __init__(self, data):
        self.data = data  # Store the value of the node
        self.children = []  # Initialize an empty list for children

# Creating a basic tree with root node 1 and two children (2 and 3)
root = TreeNode(1)  # Root node with value 1
child1 = TreeNode(2)  # Child node with value 2
child2 = TreeNode(3)  # Child node with value 3

root.children.append(child1)  # Add child1 to the root's children
root.children.append(child2)  # Add child2 to the root's children
```

```python
def print_tree_recursive(node, level=0):
    if node is None:
        return
    
    # Print the node's data with indentation based on its level
    print(str(node.data))
    
    # Recursively print all the children of the current node
    for child in node.children:
        print_tree_recursive(child, level + 2)  # Increase indentation for children

# Example usage to print the tree
print("Tree:")
print_tree_recursive(root)
```

```python
def print_tree_recursive(node):
    if node is None:
        return
    
    # Print the node's data with indentation based on its level
    print(f"{node.data}:",end=" ")
    for child in node.children:
        print(child.data,sep="",end=" ")  # Increase indentation for children

    print()
    # Recursively print all the children of the current node
    for child in node.children:
        print_tree_recursive(child)  # Increase indentation for children

# Example usage to print the tree
print("Tree:")
print_tree_recursive(root)
```

```python
def take_input_normally():
    data = int(input("Enter the data for node: "))
    node = TreeNode(data)  # Create a new node with the given data
    num_children = int(input(f"Enter the number of children for {data}: "))
    
    # Take input for each child
    for _ in range(num_children):
        child = take_input_normally()  # Recursively take input for the children
        node.children.append(child)  # Add the child to the node's children list
    
    return node  # Return the created node

# Example usage to create a tree from input
root = take_input_normally()
print("Tree Structure from Input:")
print_tree_recursive(root)
```

```python
from collections import deque

def take_input_level_wise():
    data = int(input("Enter the data for root: "))
    root = TreeNode(data)  # Create the root node
    queue = deque([root])  # Initialize a queue with the root node
    
    while queue:
        current_node = queue.popleft()  # Dequeue the front node
        
        num_children = int(input(f"Enter the number of children for {current_node.data}: "))
        
        for _ in range(num_children):
            child_data = int(input(f"Enter the data for child of {current_node.data}: "))
            child_node = TreeNode(child_data)  # Create the child node
            current_node.children.append(child_node)  # Add the child to the current node
            queue.append(child_node)  # Enqueue the child node for further input
    
    return root  # Return the root of the tree

# Example usage to create a tree from level-wise input
root = take_input_level_wise()
print("Tree Structure from Level-Wise Input:")
print_tree_recursive(root)
```

```python
class TreeNode:
    def __init__(self, data):
        self.data = data  # Data stored in the node
        self.children = []  # List of child nodes

    def add_child(self, child_node):
        self.children.append(child_node)  # Add a child node to the current node

    def print_tree(self, level=0):
        print(" " * level * 2 + str(self.data))  # Print the node with indentation
        for child in self.children:
            child.print_tree(level + 1)  # Recursively print child nodes


# Example usage
root = TreeNode("CEO")
head_of_sales = TreeNode("Head of Sales")
head_of_marketing = TreeNode("Head of Marketing")
sales_executive = TreeNode("Sales Executive")
marketing_executive = TreeNode("Marketing Executive")

root.add_child(head_of_sales)
root.add_child(head_of_marketing)
head_of_sales.add_child(sales_executive)
head_of_marketing.add_child(marketing_executive)

root.print_tree()
```

```python
def height_of_tree(node):
    if node is None:
        return 0  # Base case: if node is None, height is 0
    
    # If the node has no children, its height is 1
    if not node.children:
        return 1
    
    # Compute the height of all children and add 1 for the current node
    max_child_height = max(height_of_tree(child) for child in node.children)
    return 1 + max_child_height
```

```python
def print_nodes_at_depth(node, depth):
    if node is None:
        return
    
    # If depth is 0, print the current node's data
    if depth == 0:
        print(node.data, end=" ")
        return
    
    # Recur for all children, decreasing the depth
    for child in node.children:
        print_nodes_at_depth(child, depth - 1)

# Example usage to print nodes at depth 2
depth = 2
print(f"Nodes at depth {depth}:")
print_nodes_at_depth(root, depth)
```

```python
from collections import deque

def print_generic_tree_level_wise(root):
    if not root:
        return

    # Create a queue and enqueue the root node
    queue = deque([root])

    while queue:
        # Dequeue the front node
        current_node = queue.popleft()
        
        # Print the current node data
        print(f"{current_node.data}: ", end="")

        # Print the children of the current node
        for child in current_node.children:
            print(child.data, end=", ")
            # Enqueue the child nodes
            queue.append(child)
        print()  # For new line after each level

# Test the function with root1
print_generic_tree_level_wise(root1)
```

```python
def preorder_traversal(node):
    if node is None:
        return
    
    print(node.data, end=" ")  # Visit the current node
    
    # Recur for all the children
    for child in node.children:
        preorder_traversal(child)

# Example usage for preorder traversal
print("Preorder Traversal:")
preorder_traversal(root)
```

```python
def postorder_traversal(node):
    if node is None:
        return
    
    # Recur for all children first
    for child in node.children:
        postorder_traversal(child)
    
    print(node.data, end=" ")  # Visit the current node after its children

# Example usage for postorder traversal
print("Postorder Traversal:")
postorder_traversal(root)
```

```python
def largest_element_in_tree(node):
    if node is None:
        return float('-inf')  # Return negative infinity for an empty tree
    
    # Start with the current node's data as the largest
    largest = node.data
    
    # Recur for all children to find the maximum value
    for child in node.children:
        largest = max(largest, largest_element_in_tree(child))
    
    return largest
```

```python
def second_largest_util(node, first, second):
    if node is None:
        return first, second
    
    if node.data > first:
        second = first
        first = node.data
    elif node.data > second and node.data != first:
        second = node.data
    
    for child in node.children:
        first, second = second_largest_util(child, first, second)
    
    return first, second

def second_largest_element_in_tree(node):
    if node is None:
        return None  # Return None for an empty tree
    
    first, second = float('-inf'), float('-inf')
    first, second = second_largest_util(node, first, second)
    
    return second if second != float('-inf') else None  # Return None if no second largest exists
```

```python
def next_larger_element_util(node, number, candidate):
    if node is None:
        return candidate
    
    if node.data > number:
        if candidate is None or node.data < candidate:
            candidate = node.data
    
    for child in node.children:
        candidate = next_larger_element_util(child, number, candidate)
    
    return candidate

def next_larger_element_in_tree(node, number):
    if node is None:
        return None
    
    return next_larger_element_util(node, number, None)

# Example usage to find the next larger element
number = 5
next_larger = next_larger_element_in_tree(root, number)
print(f"Next larger element than {number}: {next_larger}")
```

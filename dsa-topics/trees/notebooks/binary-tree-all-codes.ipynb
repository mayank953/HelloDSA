```python
class BinaryTreeNode:
    def __init__(self, data):
        self.data = data  # Store the value of the node
        self.left = None  # Left child (initialized to None)
        self.right = None  # Right child (initialized to None)

# Creating a basic binary tree with root node 1 and two children (2 and 3)
root = BinaryTreeNode(1)  # Root node with value 1
root.left = BinaryTreeNode(2)  # Left child with value 2
root.right = BinaryTreeNode(3)  # Right child with value 3
```

```python
def print_binary_tree(node):
    if node is None:
        return
    
    # Format: Node: L->LeftChild, R->RightChild
    print(f"{node.data}:", end=" ")
    
    if node.left:
        print(f"L->{node.left.data}", end=", ")
    else:
        print("L->None", end=", ")
    
    if node.right:
        print(f"R->{node.right.data}")
    else:
        print("R->None")
    
    # Recur for the left and right children
    print_binary_tree(node.left)
    print_binary_tree(node.right)

# Example usage to print the tree
print("Binary Tree Structure:")
print_binary_tree(root)
```

```python
def take_input_binary_tree():
    data = int(input("Enter the data for the node: "))
    if data == -1:  # -1 means no node (base case)
        return None
    
    node = BinaryTreeNode(data)  # Create the node with given data
    
    print(f"Enter left child of {data}:")
    node.left = take_input_binary_tree()  # Recursively take input for the left child
    
    print(f"Enter right child of {data}:")
    node.right = take_input_binary_tree()  # Recursively take input for the right child
    
    return node

# Example usage to create a binary tree from input
print("Enter the binary tree data (-1 for no node):")
root = take_input_binary_tree()
print("Binary Tree Structure:")
print_binary_tree(root)
```

```python
from collections import deque

def take_input_level_wise():
    data = int(input("Enter the data for root: "))
    if data == -1:
        return None
    
    root = BinaryTreeNode(data)  # Create the root node
    queue = deque([root])  # Initialize a queue with the root node
    
    while queue:
        current_node = queue.popleft()  # Dequeue the front node
        
        left_data = int(input(f"Enter left child of {current_node.data}: "))
        if left_data != -1:
            left_node = BinaryTreeNode(left_data)
            current_node.left = left_node
            queue.append(left_node)
        
        right_data = int(input(f"Enter right child of {current_node.data}: "))
        if right_data != -1:
            right_node = BinaryTreeNode(right_data)
            current_node.right = right_node
            queue.append(right_node)
    
    return root

# Example usage to create a binary tree from level-wise input
root = take_input_level_wise()
print("Binary Tree Structure (Level-Wise Input):")
print_binary_tree(root)
```

```python
def print_level_wise(root):
    if root is None:
        return
    
    queue = deque([root])  # Initialize the queue with the root node
    
    while queue:
        current_node = queue.popleft()  # Dequeue the front node
        print(f"{current_node.data}: ", end=" ")
        
        if current_node.left:
            print(f"L->{current_node.left.data}", end=", ")
            queue.append(current_node.left)  # Enqueue the left child
        else:
            print("L->None", end=", ")
        
        if current_node.right:
            print(f"R->{current_node.right.data}")
            queue.append(current_node.right)  # Enqueue the right child
        else:
            print("R->None")

# Example usage to print the binary tree level-wise
print("Level-Wise Printing of Binary Tree:")
print_level_wise(root)
```

```python
def preorder_traversal(node):
    if node is None:
        return
    print(node.data, end=" ")  # Visit the current node
    preorder_traversal(node.left)  # Recur on left subtree
    preorder_traversal(node.right)  # Recur on right subtree

def inorder_traversal(node):
    if node is None:
        return
    inorder_traversal(node.left)  # Recur on left subtree
    print(node.data, end=" ")  # Visit the current node
    inorder_traversal(node.right)  # Recur on right subtree

def postorder_traversal(node):
    if node is None:
        return
    postorder_traversal(node.left)  # Recur on left subtree
    postorder_traversal(node.right)  # Recur on right subtree
    print(node.data, end=" ")  # Visit the current node

# Example usage to traverse the binary tree
print("Preorder Traversal:")
preorder_traversal(root)
print("\nInorder Traversal:")
inorder_traversal(root)
print("\nPostorder Traversal:")
postorder_traversal(root)
```

```python
class BinaryTreeNode:
    def __init__(self, data):
        self.data = data
        self.left = None
        self.right = None

# Utility function to calculate height and diameter
def height_and_diameter(root):
    if root is None:
        return 0, 0  # height, diameter
    
    left_height, left_diameter = height_and_diameter(root.left)
    right_height, right_diameter = height_and_diameter(root.right)
    
    current_height = 1 + max(left_height, right_height)
    diameter_through_root = left_height + right_height
    current_diameter = max(diameter_through_root, left_diameter, right_diameter)
    
    return current_height, current_diameter

def diameter_of_binary_tree(root):
    _, diameter = height_and_diameter(root)
    return diameter

# Example
root1 = BinaryTreeNode(1)
root1.left = BinaryTreeNode(2)
root1.right = BinaryTreeNode(3)
root1.left.left = BinaryTreeNode(4)
root1.left.right = BinaryTreeNode(5)

print("Diameter of the tree:", diameter_of_binary_tree(root1))
```

2. Construct Tree from Postorder and Inorder
The idea is to use the last element of postorder (the root) and split the inorder list into left and right subtrees.



```python
def build_tree_from_postorder_inorder(postorder, inorder, post_start, post_end, in_start, in_end):
    if post_start > post_end or in_start > in_end:
        return None

    root_data = postorder[post_end]  # The root is the last element in postorder
    root = BinaryTreeNode(root_data)

    inorder_index = inorder.index(root_data, in_start, in_end + 1)  # Find the root in inorder

    # Calculate the size of the left subtree
    left_subtree_size = inorder_index - in_start

    # Recursively build the left and right subtrees
    root.left = build_tree_from_postorder_inorder(postorder, inorder, post_start, post_start + left_subtree_size - 1, in_start, inorder_index - 1)
    root.right = build_tree_from_postorder_inorder(postorder, inorder, post_start + left_subtree_size, post_end - 1, inorder_index + 1, in_end)
    
    return root

# Example usage
postorder = [4, 5, 2, 3, 1]
inorder = [4, 2, 5, 1, 3]
n = len(postorder)
root = build_tree_from_postorder_inorder(postorder, inorder, 0, n - 1, 0, n - 1)
```

```python
#from predefinedBT import BinaryTreeNode, predefined_binary_tree_inputs, print_level_wise

def construct_tree_from_inorder_and_postorder(inorder, postorder, inS, inE, poS, poE):
    if inS > inE or poS > poE:  # Base Condition
        return None

    root_data = postorder[poE]  # Last element in postorder is the root
    root = BinaryTreeNode(root_data)

    rootIndexInInorder = -1
    for i in range(inS, inE + 1):
        if inorder[i] == root_data:
            rootIndexInInorder = i
            break

    if rootIndexInInorder == -1:
        print("Root not found in Inorder, please check logic")
        return None

    linS = inS
    linE = rootIndexInInorder - 1
    lpoS = poS
    lpoE = lpoS + (linE - linS)  # Left part of postorder will have the same length as the left part of inorder

    rinS = rootIndexInInorder + 1
    rinE = inE
    rpoS = lpoE + 1
    rpoE = poE - 1  # Exclude the root from postorder for the right subtree

    root.left = construct_tree_from_inorder_and_postorder(inorder, postorder, linS, linE, lpoS, lpoE)
    root.right = construct_tree_from_inorder_and_postorder(inorder, postorder, rinS, rinE, rpoS, rpoE)

    return root

# Example usage
postorder = [4, 5, 2, 6, 3, 1]
inorder = [4, 2, 5, 1, 3, 6]
n = len(inorder)
root = construct_tree_from_inorder_and_postorder(inorder, postorder, 0, n-1, 0, n-1)
print_level_wise(root)
```

3. Construct Tree from Preorder and Inorder
The first element of the preorder list is the root. We use it to split the inorder list.



```python
def build_tree_from_preorder_inorder(preorder, inorder, pre_start, pre_end, in_start, in_end):
    if pre_start > pre_end or in_start > in_end:
        return None

    root_data = preorder[pre_start]  # The root is the first element in preorder
    root = BinaryTreeNode(root_data)

    inorder_index = inorder.index(root_data, in_start, in_end + 1)  # Find the root in inorder

    # Calculate the size of the left subtree
    left_subtree_size = inorder_index - in_start

    # Recursively build the left and right subtrees
    root.left = build_tree_from_preorder_inorder(preorder, inorder, pre_start + 1, pre_start + left_subtree_size, in_start, inorder_index - 1)
    root.right = build_tree_from_preorder_inorder(preorder, inorder, pre_start + left_subtree_size + 1, pre_end, inorder_index + 1, in_end)
    
    return root

# Example usage
preorder = [1, 2, 4, 5, 3]
inorder = [4, 2, 5, 1, 3]
n = len(preorder)
root = build_tree_from_preorder_inorder(preorder, inorder, 0, n - 1, 0, n - 1)
```

4. Check If Binary Tree is Balanced
A tree is balanced if the height difference between the left and right subtrees of any node is at most 1.

```python
def is_balanced(root):
    def check_balance_and_height(root):
        if not root:
            return True, 0  # balanced, height
        
        left_balanced, left_height = check_balance_and_height(root.left)
        right_balanced, right_height = check_balance_and_height(root.right)
        
        current_balanced = left_balanced and right_balanced and abs(left_height - right_height) <= 1
        current_height = 1 + max(left_height, right_height)
        
        return current_balanced, current_height
    
    balanced, _ = check_balance_and_height(root)
    return balanced

# Example
root1.left.left = BinaryTreeNode(6)
print("Is the tree balanced?", is_balanced(root1))
```

Find Min and Max of a Binary Tree
We traverse the entire tree to find the minimum and maximum values.

```python
def find_min_max(root):
    if root is None:
        return float('inf'), float('-inf')  # Min, Max
    
    left_min, left_max = find_min_max(root.left)
    right_min, right_max = find_min_max(root.right)
    
    current_min = min(root.data, left_min, right_min)
    current_max = max(root.data, left_max, right_max)
    
    return current_min, current_max

# Example
min_val, max_val = find_min_max(root1)
print("Min:", min_val, "Max:", max_val)
```

4. Sum of All Nodes in a Binary Tree:
We can calculate the sum of all nodes using a simple postorder traversal.

```python
def sum_of_nodes(root):
    if root is None:
        return 0
    
    left_sum = sum_of_nodes(root.left)
    right_sum = sum_of_nodes(root.right)
    
    return root.data + left_sum + right_sum

# Example
root1 = BinaryTreeNode(1)
root1.left = BinaryTreeNode(2)
root1.right = BinaryTreeNode(3)
root1.left.left = BinaryTreeNode(4)
root1.left.right = BinaryTreeNode(5)

total_sum = sum_of_nodes(root1)
print("Sum of all nodes:", total_sum)
```

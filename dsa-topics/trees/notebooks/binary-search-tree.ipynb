```python
def search_bst(root, data):
    if root is None or root.data == data:
        return root  # If found, return the node (or None if not found)
    
    if data < root.data:
        return search_bst(root.left, data)
    else:
        return search_bst(root.right, data)
```

```python
def insert_bst(root, data):
    if root is None:
        return BSTNode(data)
    
    if data < root.data:
        root.left = insert_bst(root.left, data)
    else:
        root.right = insert_bst(root.right, data)
    
    return root
```

```python
# Return min max method 

class BinaryTreeNode:
    def __init__(self, data):
        self.data = data
        self.left = None
        self.right = None

# Helper function to find the maximum value in a tree
def find_max(node):
    if node is None:
        return float('-inf')  # If the node is None, return negative infinity
    left_max = find_max(node.left)
    right_max = find_max(node.right)
    return max(node.data, left_max, right_max)

# Helper function to find the minimum value in a tree
def find_min(node):
    if node is None:
        return float('inf')  # If the node is None, return positive infinity
    left_min = find_min(node.left)
    right_min = find_min(node.right)
    return min(node.data, left_min, right_min)

# Function to check if a binary tree is a BST using the max and min approach
def is_bst(node):
    if node is None:
        return True  # An empty tree is a BST
    
    # Find the maximum value in the left subtree
    left_max = find_max(node.left)
    # Find the minimum value in the right subtree
    right_min = find_min(node.right)
    
    # Check the current node's data with the max of the left subtree and min of the right subtree
    if node.data <= left_max or node.data >= right_min:
        return False
    
    # Recursively check the left and right subtrees
    return is_bst(node.left) and is_bst(node.right)

# Test the function with an example tree
def create_test_tree():
    root = BinaryTreeNode(10)
    root.left = BinaryTreeNode(5)
    root.right = BinaryTreeNode(15)
    root.left.left = BinaryTreeNode(2)
    root.left.right = BinaryTreeNode(7)
    root.right.left = BinaryTreeNode(12)
    root.right.right = BinaryTreeNode(20)
    return root

# Test the BST checker
root = create_test_tree()
if is_bst(root):
    print("The tree is a Binary Search Tree (BST).")
else:
    print("The tree is NOT a Binary Search Tree (BST).")
```

```python
def print_bst_in_range(root, low, high):
    if root is None:
        return
    
    if low < root.data:
        print_bst_in_range(root.left, low, high)
    
    if low <= root.data <= high:
        print(root.data, end=" ")
    
    if high > root.data:
        print_bst_in_range(root.right, low, high)
```

```python
class BST:
    def __init__(self):
        self.root = None
    
    # Insert a node
    def insert(self, data):
        self.root = insert_bst(self.root, data)
    
    # Search for a node
    def search(self, data):
        return search_bst(self.root, data)
    
    # Delete a node
    def delete(self, data):
        self.root = self.delete_bst(self.root, data)
    
    # Delete a node helper function
    def delete_bst(self, root, data):
        if root is None:
            return root
        
        if data < root.data:
            root.left = self.delete_bst(root.left, data)
        elif data > root.data:
            root.right = self.delete_bst(root.right, data)
        else:
            # Node with one child or no child
            if root.left is None:
                return root.right
            elif root.right is None:
                return root.left
            
            # Node with two children: get the inorder successor (smallest in the right subtree)
            min_larger_node = self.get_min_node(root.right)
            root.data = min_larger_node.data
            root.right = self.delete_bst(root.right, min_larger_node.data)
        
        return root
    
    # Helper function to get the minimum node in a subtree
    def get_min_node(self, node):
        current = node
        while current.left is not None:
            current = current.left
        return current
    
    # Check if tree is BST
    def is_bst(self):
        return is_bst(self.root)
    
    # Print BST
    def print(self):
        print_bst(self.root)
    
    # Print nodes in a range
    def print_in_range(self, low, high):
        print_bst_in_range(self.root, low, high)
```

```python
bst = BST()
bst.insert(10)
bst.insert(5)
bst.insert(20)
bst.insert(3)
bst.insert(7)

bst.print()  # Output: 3 5 7 10 20
print()

print(bst.search(7))  # Output: Node with data 7
print(bst.is_bst())   # Output: True

bst.delete(5)
bst.print()  # Output: 3 7 10 20

bst.print_in_range(4, 15)  # Output: 7 10
```

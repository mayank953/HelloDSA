```python
'''Explanation:
A fixed-size array is used, and two pointers (front and rear) are maintained.
We’ll make the queue circular, so the rear and front pointers can wrap around the array.'''
```

```python
class QueueUsingArray:
    def __init__(self, capacity):
        self.capacity = capacity  # Set the maximum capacity of the queue
        self.queue = [None] * capacity  # Initialize a fixed-size array
        self.front = -1  # Front pointer to track the front of the queue
        self.rear = -1  # Rear pointer to track the rear of the queue
        self.size = 0  # Keep track of the current size

    def is_empty(self):
        return self.size == 0  # Queue is empty if size is 0

    def is_full(self):
        return self.size == self.capacity  # Queue is full if size reaches capacity

    def enqueue(self, data):
        if self.is_full():
            print("Queue overflow. Cannot enqueue the value.")
            return
        if self.front == -1:  # First element being enqueued
            self.front = 0
        self.rear = (self.rear + 1) % self.capacity  # Circular increment
        self.queue[self.rear] = data
        self.size += 1  # Update the size
        print(f"Enqueued {data} to the queue.")

    def dequeue(self):
        if self.is_empty():
            print("Queue underflow. Cannot dequeue from an empty queue.")
            return None
        data = self.queue[self.front]
        self.front = (self.front + 1) % self.capacity  # Circular increment
        self.size -= 1  # Update the size
        if self.size == 0:  # Reset pointers when queue becomes empty
            self.front = -1
            self.rear = -1
        return data

    def front_element(self):
        if self.is_empty():
            print("Queue is empty. No front element.")
            return None
        return self.queue[self.front]  # Return the front element

    def size_of_queue(self):
        return self.size  # Return the current size
```

# Queue using List

```python
class QueueUsingList:
    def __init__(self):
        self.queue = []  # Use Python's list to represent the queue

    def is_empty(self):
        return len(self.queue) == 0  # Queue is empty if its length is 0

    def enqueue(self, data):
        self.queue.append(data)  # Append the data to the end of the list
        print(f"Enqueued {data} to the queue.")

    def dequeue(self):
        if self.is_empty():
            print("Queue underflow. Cannot dequeue from an empty queue.")
            return None
        return self.queue.pop(0)  # Remove and return the front element

    def front_element(self):
        if self.is_empty():
            print("Queue is empty. No front element.")
            return None
        return self.queue[0]  # Return the front element

    def size_of_queue(self):
        return len(self.queue)  # Return the current size of the queue
```

# Queue Using Linked List

```python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None  # Initialize the next pointer to None


class QueueUsingLinkedList:
    def __init__(self):
        self.head = None  # Head (front) of the queue
        self.tail = None  # Tail (rear) of the queue
        self.size = 0  # Keep track of the current size

    def is_empty(self):
        return self.size == 0  # Queue is empty if size is 0

    def enqueue(self, data):
        new_node = Node(data)
        if self.tail is None:  # If queue is empty
            self.head = self.tail = new_node
        else:
            self.tail.next = new_node  # Add the new node at the end
            self.tail = new_node  # Update the tail to the new node
        self.size += 1
        print(f"Enqueued {data} to the queue.")

    def dequeue(self):
        if self.is_empty():
            print("Queue underflow. Cannot dequeue from an empty queue.")
            return None
        data = self.head.data  # Get the front element
        self.head = self.head.next  # Move the head to the next element
        if self.head is None:  # If the queue becomes empty
            self.tail = None
        self.size -= 1
        return data

    def front_element(self):
        if self.is_empty():
            print("Queue is empty. No front element.")
            return None
        return self.head.data  # Return the front element

    def size_of_queue(self):
        return self.size  # Return the current size
```

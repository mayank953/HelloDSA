# Input of my LinkedList

```python
def print_LL(head):
    if(head==None):
        return
    print(head.data)
    print_LL(head.next)
```

```python
class Node:
    def __init__(self,value):
        self.data = value
        self.next = None


def take_input():
    head = None
    data = int(input("Enter the Data"))

    while(data != -1):
        newNode = Node(data)
        if(head == None):
            head = newNode
        else:
            head.next = newNode
            
        data = int(input("Enter the Data"))

    return head


def take_input_Right():
    head = None
    data = int(input("Enter the Data"))

    while(data != -1):
        newNode = Node(data)
        if(head == None):
            head = newNode
        else:
            temp = head
            while(temp!=None and temp.next !=None):
                temp = temp.next

            temp.next = newNode
            
        data = int(input("Enter the Data"))

    return head

def take_input_Best():
    head = None
    tail = None
    data = int(input("Enter the Data"))

    while(data != -1):
        newNode = Node(data)
        if(head == None):
            head = newNode
            tail = newNode
        else:
            tail.next = newNode
            tail = newNode
    
            
        data = int(input("Enter the Data"))

    return head


head = take_input_Best()
print_LL(head)
        

```

# Length - Recursion

```python
def LengthOfLinkedList(head):
    if(head == None):
        return 0

    return 1 + LengthOfLinkedList(head.next)
```

```python
def delete_tail(head):
    if(head == None) :
        return head
    
    # Only Single Element
    if(head.next == None):
        return None
        
    temp = head
    while temp.next.next is not None:
        temp = temp.next

    temp.next = None

    return head
        
```

```python
def delete_tail_recursively(head):
    if(head==None or head.next is None):
        return None

    head.next = delete_tail_recursively(head.next)
    return head
```

```python
def delete_by_value(head,value):
    if(head is None):
        return head

    if(head.data == value):
        return head.next

    temp =  head
    while(temp.next is not None and temp.next.data!=value):
        temp = temp.next

    if(temp.next is None):
        print("Value is not present")
        return head

    temp.next = temp.next.next

    return head
```

```python
def delete_by_value_recursively(head,value):
    if(head==None):
        return head

    if head.data == value:
        return head.next

    head.next = delete_by_value_recursively(head.next,value)

    return head
```

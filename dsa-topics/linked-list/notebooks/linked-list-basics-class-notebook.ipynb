```python
class Node:
    def __init__(self,value):
        self.data = value
        self.next = None
```

```python
first = Node(1)
second = Node(2)
third = Node(3)
```

```python
first.data,second.data
```

```python
id(first),id(second),id(third)
```

```python
first.next=second
second.next = third
```

```python
print(first.next)
```



# Print a LL - Recursively


```python
def print_LL(head):
    if(head==None):
        return
    print(head.data)
    print_LL(head.next)
```

```python
print_LL(head)
```

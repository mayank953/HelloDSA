```python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None  # By default, it is set to None

def take_input_better():
    head = None
    data = input("Enter the value: ")
    tail = None
    while data != "-1":
        newNode = Node(data)
        if head is None:
            head = newNode
            tail = newNode
        else:
            tail.next = newNode
            tail = newNode
        data = input("Enter the value: ")
    return head



# Example usage:
head = take_input_better()
print("Length of linked list (iteratively):", find_length_iteratively(head))
```

```python
def find_length_iteratively(head):
    length = 0 
    current = head #Explain how the same will help to perform any other operation going forward so better to save it
    while current is not None: 
        length += 1
        current = current.next
    return length

```

```python
def find_length_recursively(head):
    if head is None:
        return 0
    else:
        return 1 + find_length_recursively(head.next)
```

```python
# Search by Value
```

```python
def search_by_value(head, value):
    temp = head
    index = 0
    
    while temp is not None:  # Traverse the list
        if temp.data == value:  # If the value matches, return the index
            return index
        temp = temp.next
        index += 1
    
    return -1  # Return -1 if the value is not found
```

```python
def search_by_value_recursive(head, value, index=0):
    if head is None:
        return -1  # Base case: If list is empty or end of list is reached, return -1
    if head.data == value:
        return index  # If the value matches, return the current index
    return search_by_value_recursive(head.next, value, index + 1)  # Recur for the next node
```

```python
# Search By Index
```

```python
def search_by_index(head, index):
    temp = head
    count = 0
    
    while temp is not None:  # Traverse the list
        if count == index:  # If the current index matches the target, return the value
            return temp.data
        temp = temp.next
        count += 1
    
    return -1  # Return -1 if the index is out of bounds
```

```python
def search_by_index_recursive(head, index):
    if head is None:
        return -1  # Base case: If list is empty or end of list is reached, return -1
    if index == 0:
        return head.data  # If the index matches, return the node's data
    return search_by_index_recursive(head.next, index - 1)  # Recur with the next node and decreased index
```

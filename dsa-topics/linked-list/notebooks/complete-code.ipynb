```python
def insert_at_head(head, data):
    newNode = Node(data)
    newNode.next = head  # Point the new node's next to the current head
    head = newNode  # Update the head to the new node
    return head  # Return the updated head
```

```python
# Insert at head - recursive
```

```python
def insert_at_head_recursive(head, data):
    if head is None:
        return Node(data)  # If head is None, just return the new node
    newNode = Node(data)
    newNode.next = head
    return newNode  # Return the new node as the new head
```

```python
# Insert at tail - Iterative 
```

```python
def insert_at_tail(head, data):
    newNode = Node(data)
    if head is None:
        return newNode  # If the list is empty, the new node becomes the head
    
    temp = head
    while temp.next is not None:  # Traverse to the last node
        temp = temp.next
    temp.next = newNode  # Link the last node to the new node
    return head  # Return the original head
```

```python
def insert_at_tail_recursive(head, data):
    if head is None:
        return Node(data)  # If head is None, return the new node
    head.next = insert_at_tail_recursive(head.next, data)  # Recursively go to the last node
    return head  # Return the head after inserting at the tail
```

# Delete by Value

```python
def delete_by_value(head, value):
    if head is None:
        return None  # If the list is empty, nothing to delete
    
    if head.data == value:
        return head.next  # Special case: deleting the head if it matches the value
    
    temp = head
    while temp.next is not None and temp.next.data != value:  # Traverse to the node before the value
        temp = temp.next
    
    if temp.next is None:
        print(f"Value {value} not found")
        return head  # Return original head if the value is not found
    
    temp.next = temp.next.next  # Skip the node with the matching value
    return head  # Return the modified head
```

```python
def delete_by_value_recursive(head, value):
    if head is None:
        return None  # If the list is empty, nothing to delete
    
    if head.data == value:
        return head.next  # Delete the head if it matches the value
    
    head.next = delete_by_value_recursive(head.next, value)  # Recur for the next node
    return head  # Return the updated head
```

```python
# Search by Value
```

```python
def search_by_value(head, value):
    temp = head
    index = 0
    
    while temp is not None:  # Traverse the list
        if temp.data == value:  # If the value matches, return the index
            return index
        temp = temp.next
        index += 1
    
    return -1  # Return -1 if the value is not found
```

```python
def search_by_value_recursive(head, value, index=0):
    if head is None:
        return -1  # Base case: If list is empty or end of list is reached, return -1
    if head.data == value:
        return index  # If the value matches, return the current index
    return search_by_value_recursive(head.next, value, index + 1)  # Recur for the next node
```

```python
# Search By Index
```

```python
def search_by_index(head, index):
    temp = head
    count = 0
    
    while temp is not None:  # Traverse the list
        if count == index:  # If the current index matches the target, return the value
            return temp.data
        temp = temp.next
        count += 1
    
    return -1  # Return -1 if the index is out of bounds
```

```python
def search_by_index_recursive(head, index):
    if head is None:
        return -1  # Base case: If list is empty or end of list is reached, return -1
    if index == 0:
        return head.data  # If the index matches, return the node's data
    return search_by_index_recursive(head.next, index - 1)  # Recur with the next node and decreased index
```

```python
def middle_using_length(head):
    if head is None:
        return None  # If the list is empty, return None

    length = 0
    temp = head

    # Calculate the length of the list
    while temp is not None:
        length += 1
        temp = temp.next

    mid_index = length // 2  # Find the middle index (integer division)

    temp = head
    count = 0

    # Traverse to the middle node
    while count < mid_index:
        temp = temp.next
        count += 1
    
    return temp  # Return the middle node
```

Intuitive Example:
Imagine a race track where two runners are racing around a circular track. The fast runner (fast pointer) runs two laps for every lap the slow runner (slow pointer) runs. Eventually, the fast runner will finish the race first, but at the same moment, the slow runner will be halfway through the race. This is analogous to the fast pointer reaching the end of the list while the slow pointer is exactly at the middle.

Explanation:
The fast pointer moves two steps at a time, and the slow pointer moves one step at a time.
When the fast pointer reaches the end (or None), the slow pointer is at the middle node.


```python
def middle_using_slow_fast(head):
    if head is None:
        return None  # If the list is empty, return None

    slow = head
    fast = head

    # Move fast pointer two steps and slow pointer one step
    while fast is not None and fast.next is not None:
        slow = slow.next
        fast = fast.next.next

    return slow  # When fast reaches the end, slow will be at the middle
```

```python
def merge_two_sorted_lists(head1, head2):
    if head1 is None:
        return head2  # If first list is empty, return the second list
    if head2 is None:
        return head1  # If second list is empty, return the first list

    # Initialize a new dummy node to help with merging
    dummy = Node(0)
    current = dummy

    # Traverse both lists
    while head1 is not None and head2 is not None:
        if head1.data < head2.data:
            current.next = head1
            head1 = head1.next
        else:
            current.next = head2
            head2 = head2.next
        current = current.next

    # Append the remaining nodes of either list
    if head1 is not None:
        current.next = head1
    if head2 is not None:
        current.next = head2

    return dummy.next  # Return the merged list starting from the first node
```

```python
def reverse_list_return_head_tail(head):
    # Base case: if list is empty or has one node, return head as both head and tail
    if head is None or head.next is None:
        return head, head  # Return the head and tail (same node in case of one node)

    # Recur for the rest of the list and get the reversed head and tail
    new_head, new_tail = reverse_list_return_head_tail(head.next)

    # Reverse the current node and set the next pointer of the tail
    new_tail.next = head  # Connect the last node of reversed part to the current node
    head.next = None  # Set current node’s next to None (it becomes the new tail)

    return new_head, head  # Return the new head and the current node as the new tail
```

```python
def reverse_list_iterative(head):
    if head is None:
        return None  # If the list is empty, return None
    
    prev = None
    current = head

    while current is not None:
        next_node = current.next  # Save the next node
        current.next = prev  # Reverse the current node's pointer
        prev = current  # Move prev to the current node
        current = next_node  # Move current to the next node

    return prev  # Prev will be the new head after the loop completes
```

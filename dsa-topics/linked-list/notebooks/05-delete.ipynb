# Delete at Head

```python
def delete_at_head(head):
    if head is None:
        return None  # If the list is empty, nothing to delete
    return head.next  # Return the next node as the new head, effectively deleting the current head
```

```python
def delete_at_head_recursive(head):
    if head is None:
        return None  # If the list is empty, nothing to delete
    return head.next  # Move the head to the next node
```

```python
# Delete at Tail (Pop)
```

```python
def delete_at_tail(head):
    if head is None or head.next is None:
        return None  # If the list is empty or has one node, return None
    
    temp = head
    while temp.next.next is not None:  # Traverse to the second last node
        temp = temp.next
    temp.next = None  # Unlink the last node
    return head  # Return the modified head
```

```python
def delete_at_tail_recursive(head):
    if head is None or head.next is None:
        return None  # If the list is empty or has one node, return None
    head.next = delete_at_tail_recursive(head.next)  # Recur for the next node
    return head  # Return the updated head
```

```python
# Delete at Index (Remove)
```

```python
def delete_at_index(head, index):
    if index == 0:
        return head.next  # Special case: deleting the head
    
    temp = head
    count = 0
    
    while temp is not None and count < index - 1:  # Traverse to the node before the index
        temp = temp.next
        count += 1
    
    if temp is None or temp.next is None:
        print("Index out of bounds")
        return head  # Return original head if index is out of bounds
    
    temp.next = temp.next.next  # Skip the node at the index
    return head  # Return the modified head
```

```python
def delete_at_index_recursive(head, index):
    if index == 0:
        return head.next  # Special case: deleting the head
    if head is None or head.next is None:
        print("Index out of bounds")
        return head  # Return original head if index is out of bounds
    
    head.next = delete_at_index_recursive(head.next, index - 1)  # Recur for the next node
    return head  # Return the updated head
```

# Delete by Value

```python
def delete_by_value(head, value):
    if head is None:
        return None  # If the list is empty, nothing to delete
    
    if head.data == value:
        return head.next  # Special case: deleting the head if it matches the value
    
    temp = head
    while temp.next is not None and temp.next.data != value:  # Traverse to the node before the value
        temp = temp.next
    
    if temp.next is None:
        print(f"Value {value} not found")
        return head  # Return original head if the value is not found
    
    temp.next = temp.next.next  # Skip the node with the matching value
    return head  # Return the modified head
```

```python
def delete_by_value_recursive(head, value):
    if head is None:
        return None  # If the list is empty, nothing to delete
    
    if head.data == value:
        return head.next  # Delete the head if it matches the value
    
    head.next = delete_by_value_recursive(head.next, value)  # Recur for the next node
    return head  # Return the updated head
```

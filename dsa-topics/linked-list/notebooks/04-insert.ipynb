```python
def insert_at_head(head, data):
    newNode = Node(data)
    newNode.next = head  # Point the new node's next to the current head
    head = newNode  # Update the head to the new node
    return head  # Return the updated head
```

```python
# Insert at head - recursive
```

```python
def insert_at_head_recursive(head, data):
    if head is None:
        return Node(data)  # If head is None, just return the new node
    newNode = Node(data)
    newNode.next = head
    return newNode  # Return the new node as the new head
```

```python
# Insert at tail - Iterative 
```

```python
def insert_at_tail(head, data):
    newNode = Node(data)
    if head is None:
        return newNode  # If the list is empty, the new node becomes the head
    
    temp = head
    while temp.next is not None:  # Traverse to the last node
        temp = temp.next
    temp.next = newNode  # Link the last node to the new node
    return head  # Return the original head
```

```python
def insert_at_tail_recursive(head, data):
    if head is None:
        return Node(data)  # If head is None, return the new node
    head.next = insert_at_tail_recursive(head.next, data)  # Recursively go to the last node
    return head  # Return the head after inserting at the tail
```

```python
def insert_at_index(head, data, index):
    if index == 0:
        return insert_at_head(head, data)  # If index is 0, insert at the head
    
    newNode = Node(data)
    temp = head
    count = 0
    
    while temp is not None and count < index - 1:  # Traverse to the node before the index
        temp = temp.next
        count += 1

    ######
    # Make sure to explain this after excluding this and not directly, code fatne do
    ######
    if temp is None: 
        print("Index out of bounds")  # If index is larger than the list length
        return head
    
    newNode.next = temp.next  # Link the new node to the next node at the index
    temp.next = newNode  # Update the previous node to point to the new node
    return head
```

```python
def insert_at_index_recursive(head, data, index):
    if index == 0:
        return insert_at_head_recursive(head, data)  # Base case, insert at head
    if head is None:
        print("Index out of bounds")  # If the list ends before reaching the index
        return None
    head.next = insert_at_index_recursive(head.next, data, index - 1)  # Recur for the next node
    return head
```
